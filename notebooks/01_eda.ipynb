```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Make charts look clean and professional
sns.set_theme(style="whitegrid")

# Hardcoded absolute path so Windows never gets confused
data_path = r"D:\Personal\Vantara\vantara_cbp\data\interim\cleaned_transactions.csv"

print("⏳ Reading your clean transaction data file...")
df = pd.read_csv(data_path)

# Convert strings to actual dates so we can do math with time
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

print(f"✅ Success! Loaded {len(df)} records into the sandbox dashboard.")
```

```python
# Find the absolute last transaction date recorded in the company logs
newest_date = df['InvoiceDate'].max()
print(f"📅 Newest transaction date in data: {newest_date}")

# Group transactions by Customer ID to see how many days since their last visit
customer_summary = df.groupby('Customer ID').agg(
    Days_Since_Last_Purchase=('InvoiceDate', lambda x: (newest_date - x.max()).days)
).reset_index()

# If they haven't bought anything in over 90 days, label them as Churned (True)
customer_summary['Has_Churned'] = customer_summary['Days_Since_Last_Purchase'] > 90

# Calculate the counts and percentage
churn_counts = customer_summary['Has_Churned'].value_counts()
churn_percentages = customer_summary['Has_Churned'].value_counts(normalize=True) * 100

print("\n--- COMPANY CHURN METRICS STATUS ---")
print(f"Active Customers (Purchased within 90 days): {churn_counts[False]} ({churn_percentages[False]:.2f}%)")
print(f"Churned Customers (No purchase for > 90 days): {churn_counts[True]} ({churn_percentages[True]:.2f}%)")
```
